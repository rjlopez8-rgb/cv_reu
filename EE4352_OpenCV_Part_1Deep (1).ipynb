```python
import cv2
import numpy as np

# 1. Load the image and convert to grayscale
image = cv2.imread('blob1.png')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 2. Threshold the image to get a binary mask
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# 3. Calculate moments
moments = cv2.moments(thresh)

# 4. Calculate centroid coordinates
# Avoid division by zero by checking if M00 is greater than 0
if moments["m00"] != 0:
    cX = int(moments["m10"] / moments["m00"])
    cY = int(moments["m01"] / moments["m00"])

    # 5. Visualize the result
    cv2.circle(image, (cX, cY), 5, (0, 255, 0), -1)
    cv2.putText(image, "Centroid", (cX - 25, cY - 25), 
                cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    cv2.imshow("Centroid Detection", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    cv2.waitKey(1)
else:
    print("No blob detected.")
```

```python
import numpy as np

def rotate_numpy(image, angle_degrees):
    h, w = image.shape[:2]
    cx, cy = w / 2, h / 2
    
    # Convert angle to radians and compute trig values
    theta = np.radians(angle_degrees)
    cos_t = np.cos(theta)
    sin_t = np.sin(theta)
    
    # Create a grid of destination coordinates
    x_dest, y_dest = np.meshgrid(np.arange(w), np.arange(h))
    
    # Shift coordinates so rotation is around the center
    x_shifted = x_dest - cx
    y_shifted = y_dest - cy
    
    # Apply inverse rotation matrix
    x_src = (x_shifted * cos_t + y_shifted * sin_t + cx).astype(int)
    y_src = (-x_shifted * sin_t + y_shifted * cos_t + cy).astype(int)
    
    # Create mask for valid coordinates within image boundaries
    mask = (x_src >= 0) & (x_src < w) & (y_src >= 0) & (y_src < h)
    
    # Initialize output image
    rotated_image = np.zeros_like(image)
    rotated_image[y_dest[mask], x_dest[mask]] = image[y_src[mask], x_src[mask]]
    
    return rotated_image
```

```python
import cv2

def rotate_opencv(image, angle_degrees):
    h, w = image.shape[:2]
    center = (w / 2, h / 2)
    
    # Calculate the affine rotation matrix
    # Note: OpenCV uses clockwise rotation by default for positive angles
    # To match counter-clockwise NumPy logic, we use angle_degrees
    matrix = cv2.getRotationMatrix2D(center, angle_degrees, 1.0)
    
    # Apply the warping transformation
    rotated_image = cv2.warpAffine(image, matrix, (w, h), flags=cv2.INTER_NEAREST)
    
    return rotated_image
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the image in grayscale
img = cv2.imread('Jup03.jpg', 0)

# 2. Convert to float32 and compute the DFT
dft = cv2.dft(np.float32(img), flags=cv2.DFT_COMPLEX_OUTPUT)

# 3. Shift the zero-frequency component to the center of the spectrum
dft_shift = np.fft.fftshift(dft)

# 4. Calculate Magnitude Spectrum for visualization
magnitude_spectrum = 20 * np.log(cv2.magnitude(dft_shift[:, :, 0], dft_shift[:, :, 1]) + 1)

# 5. Apply a Low Pass Filter (LPF)
rows, cols = img.shape
crow, ccol = rows // 2, cols // 2  # center coordinates

# Create a mask: central square is 1 (pass), others are 0 (block)
mask = np.zeros((rows, cols, 2), np.uint8)
r = 30  # Radius of the filter
mask[crow-r:crow+r, ccol-r:ccol+r] = 1

# Apply mask and inverse DFT
fshift = dft_shift * mask
f_ishift = np.fft.ifftshift(fshift)
img_back = cv2.idft(f_ishift)
img_back = cv2.magnitude(img_back[:, :, 0], img_back[:, :, 1])

# 6. Normalize and display results
plt.figure(figsize=(12, 6))
plt.subplot(131), plt.imshow(img, cmap='gray'), plt.title('Input Image')
plt.subplot(132), plt.imshow(magnitude_spectrum, cmap='gray'), plt.title('Magnitude Spectrum')
plt.subplot(133), plt.imshow(img_back, cmap='gray'), plt.title('LPF Result (Blurred)')
plt.show()
```

```python
import cv2
from deepface import DeepFace

# Initialize the video capture object for the primary camera
cap = cv2.VideoCapture(0)

# Load the XML classifier for face detection (OpenCV's Haar Cascade)
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

while True:
    # Capture frame-by-frame
    ret, frame = cap.read()
    if not ret:
        break

    # Convert to grayscale for Haar Cascade detection
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    
    # Detect faces in the image
    faces = face_cascade.detectMultiScale(gray_frame, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

    for (x, y, w, h) in faces:
        # Extract the ROI (Face) for DeepFace analysis
        face_roi = frame[y:y + h, x:x + w]

        try:
            # Perform emotion analysis
            # actions=['emotion'] specifies the task
            # enforce_detection=False prevents crashing if a face is temporarily lost
            results = DeepFace.analyze(face_roi, actions=['emotion'], enforce_detection=False)

            # Extract the dominant emotion
            # results is a list of dicts for versions > 0.0.78
            dominant_emotion = results[0]['dominant_emotion']
            score = results[0]['emotion'][dominant_emotion]

            # Visualization: Draw bounding box
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

            # Display the label and confidence score
            label = f"{dominant_emotion} ({score/100:.2f})"
            cv2.putText(frame, label, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (36, 255, 12), 2)
            
        except Exception as e:
            print(f"Analysis Error: {e}")

    # Display the resulting frame
    cv2.imshow('Emotion Detection - OpenCV/DeepFace', frame)

    # Break loop on 'q' key press
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release hardware resources
cap.release()
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def analyze_square_wave(N_values, f_signal=50, fs=1000):
    """
    Analyzes a square wave using FFT for different sample counts (N).
    f_signal: Frequency of the square wave in Hz
    fs: Sampling frequency in Hz
    """
    plt.figure(figsize=(12, 8))
    
    for i, N in enumerate(N_values):
        # Time vector
        t = np.arange(N) / fs
        # Generate square wave
        x = np.sign(np.sin(2 * np.pi * f_signal * t))
        
        # Compute FFT
        X_k = np.fft.fft(x)
        freqs = np.fft.fftfreq(N, 1/fs)
        
        # Normalize magnitude
        mag = np.abs(X_k) / N
        
        # Plotting (Positive frequencies only)
        plt.subplot(len(N_values), 1, i+1)
        plt.stem(freqs[:N//2], mag[:N//2] * 2, basefmt=" ")
        plt.title(f"FFT Magnitude Spectrum (N = {N})")
        plt.ylabel("Magnitude")
        plt.grid(True)
        plt.xlim(0, 500) # Show up to 10th harmonic
        
    plt.xlabel("Frequency (Hz)")
    plt.tight_layout()
    plt.show()

# Example usage with different N values
# Higher N provides better frequency bin resolution
analyze_square_wave(N_values=[128, 512, 2048])
```

```python
import numpy as np
import matplotlib.pyplot as plt

def fourier_series_square(t, harmonics, A=1, T=2):
    """
    Synthesizes a square wave using n harmonics.
    """
    f_t = np.zeros_like(t)
    omega_0 = 2 * np.pi / T
    
    for n in range(1, harmonics + 1):
        # Only odd harmonics for a square wave
        if n % 2 != 0:
            bn = (4 * A) / (n * np.pi)
            f_t += bn * np.sin(n * omega_0 * t)
            
    return f_t

# Parameters
t = np.linspace(-2, 2, 1000)
harmonic_counts = [1, 3, 11, 51]

plt.figure(figsize=(10, 6))
for n in harmonic_counts:
    signal = fourier_series_square(t, n)
    plt.plot(t, signal, label=f"N={n} harmonics")

plt.title("Fourier Series Synthesis of a Square Wave")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

def plot_fourier_synthesis(N):
    """
    Plots the Fourier series synthesis of a square wave for N harmonics.
    """
    # Constants
    A = 1.0       # Amplitude
    T = 2.0       # Period
    omega_0 = 2 * np.pi / T
    t = np.linspace(-2, 2, 1000)
    
    # Synthesis
    f_t = np.zeros_like(t)
    for n in range(1, N + 1):
        # Square wave only contains odd harmonics
        if n % 2 != 0:
            bn = (4 * A) / (n * np.pi)
            f_t += bn * np.sin(n * omega_0 * t)
    
    # Plotting
    plt.figure(figsize=(10, 5))
    plt.plot(t, f_t, color='blue', lw=2, label=f'Sum of {N} Harmonics')
    
    # Ideal square wave for comparison
    ideal = A * np.sign(np.sin(omega_0 * t))
    plt.plot(t, ideal, 'r--', alpha=0.3, label='Ideal Square Wave')
    
    plt.title(f"Fourier Series Synthesis: N = {N}")
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude")
    plt.ylim(-1.5, 1.5)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend(loc='upper right')
    plt.show()

# Create the interactive slider
interact(
    plot_fourier_synthesis, 
    N=IntSlider(min=1, max=101, step=2, value=1, description='Harmonics:')
);
```
